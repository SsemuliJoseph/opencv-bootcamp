This script captures and displays a live video stream from a webcam or a video file using OpenCV, keeping the window open until the user presses the `Esc` key.

### 1. Imports and Source Selection

* `import cv2` and `import sys` load the OpenCV library and Python's system utilities.
* `s = 0` sets a default video source to the primary connected webcam (index `0`).
* The `if` block checks if a command-line argument was passed when running the script (such as a video file path or an alternative camera index) and assigns it to `s`.

### 2. Initialization

* `source = cv2.VideoCapture(s)` initializes the video capture object using the selected source.
* `cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)` creates a resizable display window named `Camera Preview`.

### 3. The Main Processing Loop

* `while cv2.waitKey(1) != 27:` runs a continuous loop that pauses for 1 millisecond per iteration to check for user input, terminating when the `Esc` key (ASCII value `27`) is pressed.
* `has_frame, frame = source.read()` reads the next frame from the video stream. `has_frame` is a boolean indicating success, and `frame` contains the image matrix.
* `if not has_frame: break` safely exits the loop if the video stream ends or a frame cannot be retrieved.
* `cv2.imshow(win_name, frame)` renders the current frame inside the preview window.

### 4. Resource Cleanup

* `source.release()` frees up the camera or video file handle.
* `cv2.destroyWindow(win_name)` closes the OpenCV display window and releases any remaining GUI resources.

```python
import cv2
import sys

s = 0
if len(sys.argv) > 1:
    s = sys.argv[1]

source = cv2.VideoCapture(s)

win_name = 'Camera Preview'
cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)

while cv2.waitKey(1) != 27: # Escape
    has_frame, frame = source.read()
    if not has_frame:
        break
    cv2.imshow(win_name, frame)

source.release()
cv2.destroyWindow(win_name)
```
